```python
class Employee:
    def __init__(self):
        self.name = "Rahul"            # public
        self._salary = 5000            # protected
        self.__password = "12345"      # private

    def show_details(self):
        print(self.name)
        print(self._salary)
        print(self.__password)
emp = Employee()

print(emp.name)                        # works
print(emp._salary)                     # works,but discouraged

emp.show_details()
```

```python
class Employee:
    def __init__(self):
        self.name = "Anu"            # public
        self._salary = 5000            # protected
        self.__password = "12345"      # private
    def show_parents(self):
      print("name :" , self.name)
      print("salary :" , self._salary)
      print("password :" , self.__password)
class Manager(Employee):
    def show_manager(self):
      print("name", self.name)
      print("salary", self._salary)

manager = Manager()
manager.show_manager()
```
